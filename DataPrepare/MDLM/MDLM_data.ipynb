#### 去重处理 uniprot 和 uniref 中的 peptide

```python
import sys
from Bio import SeqIO  # Biopython's module for sequence I/O
from tqdm import tqdm  # Progress bar library

# --- 配置 ---
# 将下面替换成你的实际文件名
fasta_files = ["/data1/tianang/SMILES_data/uniprotkb_length_1_TO_50_2025_04_09.fasta", "/data1/tianang/SMILES_data/uniref_length_1_TO_50_2025_04_03.fasta"]
output_file = "./Data/unique_peptide_sequences.txt"

# --- 主逻辑 ---
print("开始处理 FASTA 文件...")

unique_sequences = set() # 使用集合来自动处理重复项

# 遍历输入文件列表
for filename in fasta_files:
    print(f"\n正在处理文件: {filename}")
    try:
        # 使用 SeqIO.parse 读取 FASTA 文件，它返回一个 SeqRecord 迭代器
        # 使用 tqdm 包装迭代器以显示进度条
        # 注意：tqdm 可能无法预先知道 FASTA 文件中的确切记录数，
        # 但它仍然会显示处理的记录数和速率。
        # 使用 unit='records' 使进度条显示单位更清晰
        # record_iterator = SeqIO.parse(filename, "fasta")

        # 为了让tqdm知道总数（可选，但能让进度条更好看），可以先计数
        # 但这会读取文件两次，对于大文件可能效率不高。
        # 更常见的方法是直接迭代，tqdm会动态显示进度。
        # 我们这里直接迭代：

        # 使用 with 来确保文件句柄被正确关闭（虽然SeqIO通常会处理好）
        with open(filename, 'r') as handle:
             # 使用 tqdm 包装 SeqIO.parse
             # desc 参数为进度条添加描述性文字
            for record in tqdm(SeqIO.parse(handle, "fasta"), desc=f"读取 {filename}", unit=" 条记录"):
                # record.seq 是一个 Seq 对象，需要转换为字符串添加到集合中
                sequence_str = str(record.seq)
                unique_sequences.add(sequence_str)

    except FileNotFoundError:
        print(f"\n错误：输入文件 '{filename}' 未找到。", file=sys.stderr)
        print("请检查文件名和路径是否正确。")
        sys.exit(1) # 退出程序
    except Exception as e:
        print(f"\n处理文件 '{filename}' 时发生错误: {e}", file=sys.stderr)
        sys.exit(1) # 退出程序

total_unique_count = len(unique_sequences)
print(f"\n在所有文件中总共找到 {total_unique_count} 条独特的蛋白质序列。")

# 将独特的序列写入输出文件，并为写入过程添加进度条
try:
    print(f"\n正在将结果写入文件: {output_file}")
    with open(output_file, 'w') as outfile:
        # 包装 unique_sequences 进行迭代，以显示写入进度
        for seq in tqdm(unique_sequences, desc="写入文件", unit=" 条序列", total=total_unique_count):
            outfile.write(seq + '\n') # 每条序列写一行
    print(f"\n成功将 {total_unique_count} 条独特序列写入 {output_file}。")
    print("处理完成。")
except Exception as e:
    print(f"\n写入输出文件 '{output_file}' 时发生错误: {e}", file=sys.stderr)
    sys.exit(1)
```

#### 把这些 peptide sequence 全部转换成 SMILES string

```python
import sys
from pathlib import Path

# 添加上级目录到Python路径
# sys.path.append(str(Path(__file__).parent.parent))
from aa_seq_to_smiles import *
from rdkit import Chem
import json
from tqdm import tqdm
import pandas as pd
import numpy as np

# --- 文件配置 ---
input_peptide_file = "./Data/unique_peptide_sequences.txt"  # 包含肽序列的输入文件
output_smiles_file = "./Data/unique_peptides_as_smiles.txt" # 输出SMILES的文件
aa_smiles_dict = get_aa_smiles_dict('/data1/tianang/Projects/Synergy/DataPrepare/Data/all_aa_smiles_new_handcrafted.csv')

print(f"正在从 '{input_peptide_file}' 读取肽序列...")

# 先计算总行数以供进度条使用
with open(input_peptide_file, 'r') as f:
    total_lines = sum(1 for _ in f)

print(f"总共有 {total_lines} 条肽序列需要处理。")
print(f"正在将SMILES写入 '{output_smiles_file}'...")

# 使用 'with' 语句确保文件正确关闭
with open(input_peptide_file, 'r') as infile, open(output_smiles_file, 'w') as outfile:
    # 使用tqdm创建进度条，设置总长度和描述
    for line in tqdm(infile, total=total_lines, desc="处理进度"):
        peptide = line.strip() # 去除行首尾的空白（包括换行符）

        # 如果行不是空的
        if peptide:
            # 调用你的转换函数
            pep_obj = Peptide(peptide, aa_smiles_dict=aa_smiles_dict)
            smiles_result = Chem.MolToSmiles(pep_obj.ncTerminus_modified_mols[0])

            # 如果转换函数返回了结果 (不是None)
            if smiles_result:
                # 将结果写入输出文件，并在末尾添加换行符
                outfile.write(smiles_result + '\n')

print("处理完成。")
```
